# Industry Case Study

## E-commerce Recommendation Engine

A retail company wants to recommend products using **K-Nearest Neighbors (KNN)**.

Customer profile includes:
- Annual income
- Website visits
- Purchases
- Credit score

Without scaling, customers are matched almost entirely on **income**, producing poor recommendations.

Scaling allows every feature to contribute fairly to similarity calculations.

### Real-world algorithms that benefit from scaling
- KNN
- K-Means
- Logistic Regression
- Neural Networks
- SVM

Algorithms like Decision Trees and Random Forest generally do **not** require feature scaling.


```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler

df=pd.read_csv("../datasets/customer_scaling_dataset.csv")
df.head()
```

# Step 1 – Explore Feature Ranges

Observe how different the feature magnitudes are before scaling.


```python
print(df.describe().T[["min","max","mean","std"]])
```

```python
features=["Age","AnnualIncome","PurchasesPerYear","WebsiteVisits","CreditScore"]

fig,ax=plt.subplots(figsize=(8,4))
df[features].boxplot(ax=ax,rot=20)
plt.title("Before Scaling")
plt.tight_layout()
plt.show()
```

```python
X=df[features]

standard=StandardScaler().fit_transform(X)
minmax=MinMaxScaler().fit_transform(X)
robust=RobustScaler().fit_transform(X)

std_df=pd.DataFrame(standard,columns=features)
mm_df=pd.DataFrame(minmax,columns=features)
rb_df=pd.DataFrame(robust,columns=features)

print("StandardScaler")
print(std_df.head())
```

```python
fig,ax=plt.subplots(figsize=(8,4))
std_df.boxplot(ax=ax,rot=20)
plt.title("After Standard Scaling")
plt.tight_layout()
plt.show()
```

# Comparison

## StandardScaler
- Mean becomes approximately 0
- Standard deviation becomes 1
- Good for normally distributed data

## MinMaxScaler
- Scales values between 0 and 1
- Useful for neural networks and image features

## RobustScaler
- Uses median and IQR
- Best when outliers are present

### Banking Example
Credit-card fraud detection often contains extreme transaction amounts. RobustScaler performs better because it is less affected by outliers.


# Best Practices

- Scale **after** train-test split to avoid data leakage.
- Fit the scaler only on the training data.
- Use the same fitted scaler during inference.
- Save the scaler with the trained model (pickle/joblib).
