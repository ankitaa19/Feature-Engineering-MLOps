```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer

df = pd.read_csv('../../datasets/customer_loan_missing_values.csv')
df.head()
```

```python
df.info()

print("\nMissing Values")
print(df.isnull().sum())

missing_percentage = (df.isnull().sum() / len(df)).round(2)

print("\nMissing Percentage")
```

```python
missing_percentage.plot(kind="bar")

plt.title("Percentage of Missing Values")
plt.ylabel("% Missing")
plt.tight_layout()
plt.show()
```

```python
numeric=["Age","MonthlyIncome","CreditScore"]

median_imp = SimpleImputer(strategy='median')
df[numeric] = median_imp.fit_transform(df[numeric])

cat=["City"]
mode_imp = SimpleImputer(strategy='most_frequent')
df[cat] = mode_imp.fit_transform(df[cat])

print(df.isnull().sum())
```
